# Occorrenze GBIF della Svizzera

## Istruzioni

Apri Jupyter dalla cartella principale del progetto Clima, seleziona l'ambiente Python del progetto (`.venv`) ed esegui le celle dall'alto verso il basso.

Sono disponibili due funzioni. `scarica_gbif_svizzera_ricerca(limite_righe=1000)` usa l'API di ricerca senza SQL: limita il numero di righe, restituisce tutte le colonne e non richiede credenziali; è l'opzione rapida per provare. `scarica_gbif_svizzera_sql(limite_righe=1000)` usa i download SQL asincroni: consente di scegliere le colonne in `src/util/gbif_svizzera.sql` oppure passare una query personalizzata con `sql=...`; richiede credenziali GBIF e può essere lenta perché GBIF elabora la query. Se passi `sql=...`, la query deve iniziare con `SELECT` e leggere dalla tabella `occurrence`; `limite_righe` sostituisce un eventuale `LIMIT` finale.

Per i download SQL, la funzione legge dal file `.env` nella root del progetto (`clima/.env`) `GBIF_USERNAME`/`GBIF_PASSWORD` oppure `GBIF_USER`/`GBIF_PWD`. Il file `.env` è ignorato da Git.

La cella seguente usa la ricerca senza SQL. I risultati restano in memoria nel kernel Jupyter; per salvarli facoltativamente: `df_gbif_ch.to_csv("gbif_svizzera.csv", index=False)`. Per provare la variante SQL basata sul template, esegui `df_gbif_ch_sql = scarica_gbif_svizzera_sql(limite_righe=1000)`. La cella SQL più sotto mostra come passare una query personalizzata.

```python
import importlib
from pathlib import Path
import sys

# Rende importabili i moduli del progetto anche senza installarlo con pip -e .
project_src = Path.cwd() / "src"
if not (project_src / "util" / "gbif_svizzera.py").is_file():
    raise FileNotFoundError("Apri Jupyter dalla cartella principale del progetto Clima.")
sys.path.insert(0, str(project_src.resolve()))

# Ricarica il modulo per evitare di usare una versione rimasta in memoria nel kernel.
import util.gbif_svizzera as gbif_svizzera
gbif_svizzera = importlib.reload(gbif_svizzera)
scarica_gbif_svizzera_ricerca = gbif_svizzera.scarica_gbif_svizzera_ricerca
scarica_gbif_svizzera_sql = gbif_svizzera.scarica_gbif_svizzera_sql
```

```python
df_gbif_ch = scarica_gbif_svizzera_ricerca(limite_righe=1000)

print(f"Occorrenze ricevute: {len(df_gbif_ch):,}")
display(df_gbif_ch.head())
```

```python
# Query SQL personalizzabile; LIMIT viene impostato da limite_righe nella funzione.

sql_gbif_ch = """SELECT 
  gbifID, 
  datasetKey, 
  species, 
  speciesKey, 
  decimalLatitude, 
  decimalLongitude, 
  coordinateUncertaintyInMeters, 
  basisOfRecord, 
  "year", 
  "month", 
  eventDate
FROM occurrence
WHERE speciesKey = 2435099 AND hasCoordinate = true AND countryCode = 'CH'"""

df_gbif_ch_sql = scarica_gbif_svizzera_sql(
    limite_righe=1000,
    sql=sql_gbif_ch,
)
print(f"Occorrenze SQL ricevute: {len(df_gbif_ch_sql):,}")
display(df_gbif_ch_sql.head())
```

```python
# Opzionale: salva i risultati ricevuti in un CSV nella cartella del progetto.
df_gbif_ch.to_csv("gbif_svizzera.csv", index=False)



df_gbif_ch.head(10).T

for e in df_gbif_ch.columns:
    print(f"{e}: {df_gbif_ch[e].dtype}")
```
