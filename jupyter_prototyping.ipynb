# Download GBIF Svizzera

## Istruzioni

1. Apri Jupyter dalla cartella principale del progetto Clima e seleziona l'ambiente Python del progetto (`.venv`).
2. Se non hai ancora installato il progetto nell'ambiente, esegui nel terminale `python -m pip install -e .`.
3. Esegui le celle dall'alto verso il basso. Il notebook carica `GBIF_USER`, `GBIF_PWD` e `GBIF_EMAIL` dal file `.env` nella cartella principale. Non condividere né caricare online il file `.env`.
4. La funzione `scarica_gbif_svizzera` invia la richiesta a GBIF, attende che il download sia pronto e restituisce un DataFrame Pandas con i record associati al codice paese `CH`. Puoi passare `limite_righe=1000` per ottenere al massimo 1.000 righe nel DataFrame. Il limite viene applicato quando si legge il file: GBIF prepara e trasferisce comunque l'archivio completo.

Il download può richiedere tempo. Al termine, il DataFrame è disponibile nella variabile `df_gbif_ch`: puoi esplorarlo con `df_gbif_ch.head()` e salvarlo in CSV con `df_gbif_ch.to_csv("gbif_svizzera.csv", index=False)`.

```python
import os
from pathlib import Path
import sys
from dotenv import load_dotenv

# Rende importabili i moduli del progetto anche se non è stato installato con pip -e .
project_root = Path.cwd()
project_src = project_root / "src"
if not (project_src / "util" / "gbif_svizzera.py").is_file():
    raise FileNotFoundError("Apri Jupyter dalla cartella principale del progetto Clima.")
load_dotenv(dotenv_path=project_root / ".env")
sys.path.insert(0, str(project_src.resolve()))

from util.gbif_svizzera import scarica_gbif_svizzera
```

```python
gbif_username = os.getenv("GBIF_USER")
gbif_password = os.getenv("GBIF_PWD")
gbif_email = os.getenv("GBIF_EMAIL")

variabili_mancanti = [
    nome
    for nome, valore in {
        "GBIF_USER": gbif_username,
        "GBIF_PWD": gbif_password,
        "GBIF_EMAIL": gbif_email,
    }.items()
    if not valore
]
if variabili_mancanti:
    raise RuntimeError(
        "Variabili mancanti nel file .env: " + ", ".join(variabili_mancanti)
    )

print("Credenziali GBIF caricate da .env.")
```

```python
df_gbif_ch = scarica_gbif_svizzera(
    username=gbif_username,
    password=gbif_password,
    email=gbif_email,
    limite_righe=1000,
)

print(f"Occorrenze scaricate: {len(df_gbif_ch):,}")
display(df_gbif_ch.head())
```
